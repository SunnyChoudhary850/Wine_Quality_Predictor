```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
from sklearn.metrics import accuracy_score
```

```python
data = pd.read_csv("/content/winequality-red.csv")
```

```python
data
```

```python
data.info()
```

```python
X = data.iloc[:,0:-1]
Y = data.iloc[:,-1]
```

```python
# *********** TESTING A DATA SAMPLE **********
#test_a=[8.2,0.760,0.1,2.3,0.092,15.0,54.0,0.99700,3.26,0.65,9.0]
#test_a = np.array(test_a).reshape(1, -1)  # shape: (1, 11)
#test_a = scaler.transform(test_a)
#test_a
```

```python
train_x,test_x,train_y,test_y = train_test_split(X,Y,test_size=0.2,random_state=0)
```

```python
train_x=scaler.fit_transform(train_x)
```

```python
test_x=scaler.transform(test_x)
```

```python

#  LogisticRegression to train the model
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
clf=LogisticRegression()
clf.fit(train_x, train_y)
```

```python
y_pred = clf.predict(test_x)
```

```python
y_pred
```

```python
# *********** RESULT OF THE TESTING THE DATA SAMPLE **********
#a_pred = clf.predict(test_a)
#a_pred
```

```python
accuracy_score(test_y,y_pred)
```

```python
# RandomForestClassifier to train the model
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, roc_auc_score
```

```python
rf = RandomForestClassifier(n_estimators=100, class_weight='balanced', max_depth=10, random_state=42)
```

```python
rf.fit(train_x, train_y)
```

```python
test_x
```

```python
pred_yy = rf.predict(test_x)
```

```python
pred_yy
```

```python
print(classification_report(test_y, pred_yy))
```

```python
accuracy = accuracy_score(test_y, pred_yy)
print("Accuracy:", accuracy)
```

```python
# ********** DecisionTreeClassifier **********
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
# Create decision tree classifier
dt = DecisionTreeClassifier(max_depth=5, random_state=0)
dt.fit(train_x, train_y)
```

```python
pred_yyy = dt.predict(test_x)
print("Accuracy:", accuracy_score(test_y, pred_yyy))

# Detailed report
print(classification_report(test_y, pred_yyy))
```

```python
from sklearn import tree

dt = DecisionTreeClassifier(max_depth=5, random_state=42)
dt.fit(train_x, train_y)

plt.figure(figsize=(15,10))
tree.plot_tree(dt, feature_names=X.columns, class_names=[str(c) for c in dt.classes_], filled=True)
plt.show()
```
